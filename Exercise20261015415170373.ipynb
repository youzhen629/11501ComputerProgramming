習題一：使用random模組，產生出一個介於0到1之間的小數$r$，使用while迴圈，計算出$r$的前二十位二進位表示$r = 0.b_1b_2b_3\cdots {b_{20}}_2$。輸出格式如附圖Exercise20261015.png
import random

r = random.random()

bits = ""
x = r

i = 0
while i < 20:
    x = x * 2

    if x >= 1:
        bits += "1"
        x = x - 1
    else:
        bits += "0"

    i += 1

print(f"r = {r}")
print(f"r ≈ 0.{bits}₂")

Python 的 `int` 型別採用 Arbitrary-Precision Integer（任意精度整數）設計，能夠精確表示任意大小的整數，其實作方式可參考 CPython 的 `PyLongObject`。與 C/C++ 的 `int` 或 `long long` 不同，Python 的整數不受固定 32 位元或64 位元限制，而是會依據數值大小自動配置更多記憶體，因此只要記憶體足夠，便能表示極大的整數。例如：
```python
n = 10**68
print(n)
```
上述程式可正常執行，表示 Python 能夠精確儲存$10^{68}$，即中文數字中的「無量大數」。


可利用下列程式觀察不同大小整數所需的位元數（bit）與記憶體空間：

```python
import sys
n = 10**68
print(f"bit_length = {n.bit_length()}")
print(f"memory = {sys.getsizeof(n)} bytes")
```

small_unit = ["", "拾", "佰", "仟"]

big_unit = [
    "", "萬", "億", "兆", "京", "垓", "秭", "穰",
    "溝", "澗", "正", "載", "極",
    "恆河沙", "阿僧祇", "那由他",
    "不可思議", "無量大數"

def convert_four(num_str):
    """處理四位數以內"""
    result = ""

    length = len(num_str)

    for i, ch in enumerate(num_str):
        n = int(ch)

        pos = length - i - 1

        if n != 0:
            result += digit[n] + small_unit[pos]
        else:
            if result and not result.endswith("零"):
                result += "零"

    result = result.rstrip("零")

    return result


n = int(input())

if n == 0:
    print("零")
else:

    s = str(n)

    groups = []

    while s:
        groups.insert(0, s[-4:])
        s = s[:-4]

    answer = ""

    total = len(groups)

    for i, g in enumerate(groups):

        part = convert_four(g)

        if part:
            answer += part
            answer += big_unit[total - i - 1]

    print(answer)
