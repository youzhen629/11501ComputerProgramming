## 第三部分：CPython 如何實作這些型別？

除了知道 Python 使用哪些型別表示數學物件之外，

我們也希望了解：

> Python 物件在 CPython 中是如何實作的？

請查閱資料，完成下表。

| 數學概念 | Python 型別 | CPython 實作 | 主要特點 |
|----------|------------|-------------|----------|
| 自然數 | `int` | PyLongObject | 可表示任意大小整數 |
| 整數 | `int` | PyLongObject | 正負整數共用同一型別 |
| 實數 | `float` | PyFloatObject | 底層使用 C 的 double，近似表示實數 |
| 複數 | `complex` | PyComplexObject | 儲存實部與虛部 |
| 集合 | `set` | PySetObject | 元素不重複，查找速度快（雜湊表） |
| 命題真假 | `bool` | PyBoolObject | 只有 True 與 False，為 int 的子型別 |
| 串列 | `list` | PyListObject | 可變動、可儲存不同型別資料 |

## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)
```

回答：

1. 為什麼 Python 能表示如此大的整數？
因為 Python 的 int 採用**任意精度整數（Arbitrary Precision Integer）**實作，不受 CPU 32 位元或 64 位元整數大小限制。
2. Python 的 `int` 是否具有固定的位元數限制？
否。
3. 什麼是 Arbitrary Precision（任意精度）？
整數可依需要使用任意多位數來表示，而不是固定 32 或 64 位元。
4. CPython 使用什麼物件來實作 `int`？
PyLongObject
5. Python 整數大小真正的限制是什麼？
理論上沒有固定位數限制。

實際上的限制來自：

可用記憶體（RAM）
作業系統限制
Python 物件大小限制
---

再執行：

```python
0.1 + 0.2
```

```python
0.1 + 0.2 == 0.3
```

```python
format(0.1, ".20f")
```

回答：

1. 為什麼結果與數學上的答案不同？
因為 float 使用二進位近似表示小數，所以會有誤差。
2. 浮點數是否能精確表示所有小數？
不能，許多小數（如 0.1）只能近似表示。
3. 什麼是浮點數的精度（precision）？
精度是能保留的有效數字位數。
4. 什麼是浮點數的範圍（range）？
範圍是能表示的最大與最小數值。
5. 精度與範圍有何不同？
精度關心「準不準」，範圍關心「大不大」。

## 第六部分：AI 模型中的數值表示

人工智慧模型需要儲存大量參數（Parameters）與權重（Weights）。

請查閱資料並回答：

1. AI 模型中的權重通常使用什麼型別儲存？
AI 權重通常使用浮點數（FP32、FP16、BF16）。
2. 什麼是 FP64？
FP64：64 位元雙精度浮點數。
3. 什麼是 FP32？
FP32：32 位元單精度浮點數。
4. 什麼是 FP16？
FP16：16 位元半精度浮點數。
5. 什麼是 BF16？
BF16：16 位元浮點數，範圍接近 FP32。
6. 為什麼大型 AI 模型不直接使用 Python 的 `int` 作為權重？
因為權重需要表示小數，且浮點數更省記憶體並支援 GPU 加速，所以不使用 Python 的 int。

## 第八部分：綜合思考

### 思考題一

Python 官方文件指出：

```python
int
```

具有 Unlimited Precision（無限制精度）。

請回答：

1. Unlimited Precision 是什麼意思？
整數大小不受固定 32/64 位元限制
2. 它是否代表可以儲存無限大的整數？
不代表真正無限大，仍受記憶體與系統資源限制
3. CPython 如何實現這項特性？
CPython 使用PyLongObject，透過增加數字區塊的方式動態擴充儲存空間
---

### 思考題二

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？
因為 Python 的 int 採用**任意精度（Arbitrary / Unlimited Precision）**設計。
2. 為什麼 `float` 卻存在範圍限制？
float 有範圍限制，因為只使用固定的 64 位元儲存。
3. 為什麼 `float` 可能產生誤差？
float 可能有誤差，因為許多十進位小數無法用有限的二進位位元精確表示
---

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。
Python int 強調正確性；FP16/BF16 強調效率。
2. 各自的優點與缺點。
int

優點：精確、無固定上限
缺點：較耗記憶體、較慢

FP16/BF16

優點：省記憶體、速度快
缺點：有誤差、精度較低
3. 為什麼大型語言模型通常選擇較低精度的表示方式？
大型語言模型使用低精度表示，是為了節省記憶體並提高訓練與推論速度。
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？
若全部改用 FP64，會造成記憶體需求暴增、訓練變慢、可訓練模型變小，以及 GPU 效率下降。

### 挑戰題

請用自己的話解釋：

```python
x = 10**1000
```

為什麼 Python 可以正常運作；

而大型 AI 模型卻通常選擇使用：

```text
FP32
FP16
BF16
```

而不是任意精度整數或高精度浮點數？

請從：

- 記憶體使用量
- 計算速度
- 數值精度
- GPU 運算效率

等角度進行討論。
Python 的 int 追求「算得完全正確」；而大型 AI 模型追求「在可接受的誤差下，用最少記憶體與最快速度完成運算」，所以通常選擇 FP32、FP16 或 BF16，而不是任意精度整數或高精度浮點數。
